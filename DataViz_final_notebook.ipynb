```python
##alternative 
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import plotly.io as pio

# ---------------------------
# Global configuration
# ---------------------------
pio.renderers.default = "notebook"

GLOBAL_FONT = "Inter, 'Segoe UI', Roboto, Helvetica, Arial, sans-serif"
BG_COLOR = "#f0f2f5"
TEXT_COLOR = "#1a202c"
SUB_TEXT_COLOR = "#718096"

AREA_STYLE = {
    "Out of Area":        "#0077BB",  # Clear blue
    "Industrial":         "#CC3311",  # Vermillion red
    "Downtown/IU Campus": "#332288",  # Indigo
    "Residential":        "#EE7733",  # Orange
    "Suburban Towns":     "#AA3377",  # Magenta
    "Rural Towns":        "#DDCC77"   # Sand/tan
}

SERVICE_CATEGORY_COLORS = {
    "Environmental Issues":          "#117733",  # Green
    "Construction & Infrastructure": "#DDAA33",  # Gold
    "Public Safety":                 "#CC3311",  # Vermillion
    "City Services & Administrative":"#AA3377",  # Magenta
    "Transportation and Street Operations": "#0077BB",  # Clear blue
    "Utilities":                     "#332288"   # Indigo
}

seasons = ['Winter', 'Spring', 'Summer', 'Fall']

# ---------------------------
# Example Data Preparation
# ---------------------------
# Ensure df_small exists and has these columns: 'service_category', 'area_category', 'service_name', 'updated_datetime'
df_small['service_category'] = df_small['service_category'].replace({
    'Environment Issuess': 'Environmental Issues'
})
df_small['updated_datetime'] = pd.to_datetime(df_small['updated_datetime'])

# Extract season
def get_season(date):
    month = date.month
    if month in [12,1,2]: return 'Winter'
    elif month in [3,4,5]: return 'Spring'
    elif month in [6,7,8]: return 'Summer'
    else: return 'Fall'

df_small['season'] = df_small['updated_datetime'].apply(get_season)

# Encode service categories for Parcats color
df_small['service_category_code'] = df_small['service_category'].map(
    lambda x: list(SERVICE_CATEGORY_COLORS.keys()).index(x) if x in SERVICE_CATEGORY_COLORS else -1
)

# ---------------------------
# Helper functions
# ---------------------------
def get_area_color(name): 
    return AREA_STYLE.get(name, "#a0aec0")

def get_cat_color(name): 
    return SERVICE_CATEGORY_COLORS.get(name, "#cbd5e0")

def get_sunburst_data(df, season):
    df_s = df[df['season']==season]
    df_grouped = df_s.groupby(['area_category','service_category']).size().reset_index(name='count')
    ids, labels, parents, values, colors = [], [], [], [], []

    # Inner ring (area)
    area_stats = df_grouped.groupby('area_category')['count'].sum().reset_index()
    for _, row in area_stats.iterrows():
        area = row['area_category']
        ids.append(f"{season}-{area}")
        labels.append(area)
        parents.append("")
        values.append(row['count'])
        colors.append(get_area_color(area))

    # Outer ring (category)
    for _, row in df_grouped.iterrows():
        area, cat = row['area_category'], row['service_category']
        ids.append(f"{season}-{area}-{cat}")
        labels.append(cat)
        parents.append(f"{season}-{area}")
        values.append(row['count'])
        colors.append(get_cat_color(cat))

    return ids, labels, parents, values, colors

def get_parcat_data(df, season):
    df_s = df[df['season']==season]
    # Use these as dimensions: Area -> Category -> Service
    dims = [
        dict(label="Area", values=df_s['area_category'].tolist()),
        dict(label="Category", values=df_s['service_category'].tolist()),
        dict(label="Service", values=df_s['service_name'].tolist())
    ]
    return dims

# ---------------------------
# Build Figure
# ---------------------------
fig = make_subplots(
    rows=1, cols=2,
    column_widths=[0.45,0.55],
    specs=[[{"type":"sunburst"},{"type":"parcats"}]],
    subplot_titles=("Quality of Life by Area & Category","Request Flow by Service")
)

# Add traces for each season (initially visible only for first)
for i, season in enumerate(seasons):
    # Sunburst
    sb_ids, sb_lbls, sb_pars, sb_vals, sb_cols = get_sunburst_data(df_small, season)
    fig.add_trace(
        go.Sunburst(
            ids=sb_ids,
            labels=sb_lbls,
            parents=sb_pars,
            values=sb_vals,
            branchvalues="total",
            marker=dict(colors=sb_cols, line=dict(color='white', width=1.5)),
            textinfo="label+percent parent",
            hoverinfo="label+percent parent",
            name=season,
            visible=(i==0)
        ),
        row=1, col=1
    )

    # Parcats
    parcat_dims = get_parcat_data(df_small, season)
    # Line color based on service_category_code
    line_color = df_small[df_small['season']==season]['service_category_code'].tolist()
    colorscale = [[i/(len(SERVICE_CATEGORY_COLORS)-1), c] for i, c in enumerate(SERVICE_CATEGORY_COLORS.values())]

    fig.add_trace(
        go.Parcats(
            dimensions=parcat_dims,
            line=dict(color=line_color, shape="hspline", colorscale=colorscale),
            visible=(i==0)
        ),
        row=1, col=2
    )

fig.update_layout(
    annotations=[
        dict(
            text="Sunburst diagram of Distribution of service request categories across neighborhoods",
            x=0.225, y=0.95,  # slightly below the title
            xref="paper", yref="paper",
            showarrow=False,
            font=dict(size=16, color="#4a5568")
        ),
        dict(
            text="Sankey Flow from categories to specific services in Regions",
            x=0.725, y=0.95,  # slightly below the title
            xref="paper", yref="paper",
            showarrow=False,
            font=dict(size=16, color="#4a5568")
        )
    ]
)

# ---------------------------
# Dropdown for Seasons
# ---------------------------
buttons=[]
for i, season in enumerate(seasons):
    flags = [False]*(len(seasons)*2)
    flags[i*2] = True
    flags[i*2+1] = True
    buttons.append(dict(label=season, method="update", args=[{"visible":flags}]))

fig.update_layout(
    title=dict(text="<b>How do seasonal service demands impact the quality of life for Bloomington’s residents?</b>", font=dict(size=24, family=GLOBAL_FONT), x=0.5),
    updatemenus=[dict(active=0, buttons=buttons, x=0.95, y=1.12, xanchor='right', yanchor='top')],
    template="plotly_white",
    paper_bgcolor=BG_COLOR,
    plot_bgcolor=BG_COLOR,
    font=dict(family=GLOBAL_FONT, color=TEXT_COLOR),
    height=900,
    width=1500
)

# Export to HTML
fig.write_html("Bloomington_Seasonal_QoL.html")
```
